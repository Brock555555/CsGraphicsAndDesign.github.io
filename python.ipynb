Variables are Objects

An object represents a value that is automatically created by the interpreter.
An object can have more than one name assigned to it, this would be variable names, also known as Name Binding

There are 5 main properties of objects

1. Value (string, integer)
2. Type - determines mutability (whether or not an objects value is allowed to be changed)
3. Identity (telling objects apart)
4. State (what attributes an object has)
5. Behavior (what methods it can use)

Trying to change an immutable objects results in a new object being created.

```python
x = 2
print(type(x))
print(id(x))

y = 2
print(id(y))

print(y == x)

z = "hello"
print(type(z))
print(id(z))

v = z.upper()
print(id(v))

print(z == v)
```

Everything makes more sense when we create our own class

```python
class toy:
    type = 11 #this is a classes attributes, only accessible by the class itself
    def __init__(self, type, danger): #this is the initialization method or constructor
        """
        __init__() method is a constructor which is automatically called when a new object of a class 
        is created. Its main purpose is to initialize the object’s attributes and set up its initial state.
        When an object is created, memory is allocated for it and __init__ helps organize that memory 
        by assigning values to attributes. It can use default values.
        """
        #these are your objects attributes, they are what allows you to save information in reference to an object
        self.type = type
        self.danger = danger

        """
        self refers to the current object
        type and danger are parameters
        self.type and self.danger are attributes that store these values
        
        """

    def value(self): #this is a method of the class
        return 11

    def value3(self): #this is a method of the class, will be inherited
            return 33

    #this is a class method, which objects will inheret and can use unless otherwise specified
    def toy_info(self):
        return f'{self.type}, danger: {self.danger}'

    @staticmethod #a decorator to make a static method
    def static_method():
        return toy.type #a method for the class itself, callable without an object

    @classmethod #creating a factory method (creating objects of a type by using the class itself)
    def from_cartesian(cls, type, danger):
        return cls(type, danger)

toy1 = toy("Ninja_Star", 10)
toy2 = toy("Baseball_Bat", 8)

print(toy1.type, toy1.danger)
print(toy2.type, toy2.danger)

print(toy1.toy_info())
print(toy2.toy_info())

print(toy1.type)
print(toy.type)

print(toy.static_method())

p = toy.from_cartesian("ball", 2)
print(p.type, p.danger)

class balls(toy):#balls is a child class of toy
    def __init__(self, type, danger):
        super().__init__(type, danger) #super tells the class to use the constructor from the parent class

    def value(self):
        return 12

    def value2(self):
        return super().value()

ball1 = balls("Soccer", 3)
print(toy1.value())
print(ball1.value()) #as we can see the child class will use its own value unless otherwise stated
print(ball1.value2())
print(ball1.value3())
```

Objects can store attributes, which are the values that differentiate one object of a class from another. Different people have different names, different strings contain different characters, and so on.

Classes can store attributes, which are values that are equally applicable to all objects of the same class, which is to say that they describe the type rather than each object of that type.

Methods are stored as class attributes, but we call them on individual objects, which means that a method is a way to ask an object to do some job. The self parameter gives us a way to interact with the object that a method was called on. 

This enables a concept called Encapsulation, which involves bundling data (attributes) and methods into a single unit (object). This allows direct access to some of the objects components, and restrictions from others. Using classes we can make certain attributes or methods public or private to ensure certain things are or are not allowed to be called.

```python
class Person:
    #Public variables are accessible from anywhere in the program. By default, 
    #all class members in Python are public.
    def __init__(self, name, age):
        self.name = name # Public variable
        self.age = age # Public variable

    #Protected variables are intended to be accessed only within the class and its subclasses. 
    #They are prefixed with a single underscore _. 
    #However, this is a convention and not strictly enforced by Python.

    def __init__(self, name, age):
        self._name = name # Protected variable
        self._age = age # Protected variable

    #Private variables are strictly accessible only within the class.
    #They are prefixed with a double underscore __. 
    #Python uses name mangling to make these variables harder to access from outside the class.

    def __init__(self, name, age):
        self.__name = name # Private variable
        self.__age = age # Private variable
```

```python
#Python apparently also has C structs
from dataclasses import dataclass

@dataclass
class Employee:
    name: str
    dept: str
    salary: int

john = Employee('john', 'computer lab', 1000)
print(john.dept)
print(john.salary)
```
